# Problem statement.
### Build A Machine learning model to accuretely predict whether or not the patients in the dataset have diabetes or not?

## 2. Data Collection

### Import Necessary libraries.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,precision_score,confusion_matrix,classification_report
from keras.layers import Dense,Dropout
from keras.models import Sequential
import numpy as np
from keras.models import load_model
import warnings
warnings.filterwarnings("ignore")
```

### Import data set.

```python
diabetes_data = pd.read_csv("diabetes.csv")
diabetes_data
```

## 3. Data Understanding

```python
diabetes_data.shape
```

```python
diabetes_data.isna().sum()
```

```python
diabetes_data.dtypes
```

```python
diabetes_data.describe()
```

## 4. Data Preparation

```python
X = diabetes_data.drop(labels='Outcome',axis = 1)
y = diabetes_data[["Outcome"]]
```

```python
X.shape
```

```python
y.shape
```

## 5. Model Building

```python
model = Sequential()
model.add(Dense(12,input_dim = 8,kernel_initializer='uniform',activation='relu'))
model.add(Dense(8,kernel_initializer='uniform',activation='relu'))
model.add(Dense(1,kernel_initializer='uniform',activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
X_train,X_test,y_train,y_test =  train_test_split(X,y,test_size=0.20,random_state=12,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

### Alert: Here's where you Need to apply Standarization.

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)
```

## 6. Model Training || 7. Model Testing || 8. Evaluation

```python
model.fit(X_train,y_train,batch_size=16,epochs=50,validation_split=0.2)
```

## 9. Model Prediction.

```python
y_pred = model.predict(X_test)
y_pred = (y_pred > 0.50).astype(int)
y_pred
```

```python
print("Accuracy_Score:",round(accuracy_score(y_test,y_pred),4))
print("precision_Score:",round(precision_score(y_test,y_pred),3))
print("confusion_matrix:\n",confusion_matrix(y_test,y_pred))
print("Classification Report:",classification_report(y_test,y_pred))
```

```python
model.save("Healthcare_intelligence.h5")
```
